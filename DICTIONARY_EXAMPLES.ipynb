```python
#METHODS DICTIONARY
```

```python
#1)GET():
student={'rollno':101,'name':"dhanashree",'cname':'DS'}
print(student.get("name"))
print(student.get("cname"))

```

```python
#2)KEYS():ALL KEYS
student.keys()
```

```python
#3)VALUES():ALL VALUES
student.values()
```

```python
#4)ITEMS():KEYS VALUES PAIRS
student={'rollno':101,'name':"dhanashree",'cname':'DS'}
student["city"]="pune"
print(student)
print(len(student))
```

```python
#5)dictionary delete item delete
student={'rollno':101,'name':"dhanashree",'cname':'DS'}
del student["cname"]
print(student)
```

```python
#6)UPADTE():update  dictionary
student={'rollno':101,'name':"dhanashree",'cname':'DS', "age":21}
student.update({"age":23})
print(student)
```

```python
#7)POP()=SPECIFIC KEY VALUE DELETE
Student={'rollno':101,'name':"dhanashree",'cname':'DS', "age":21}
student.pop("age")
print(student)
```

```python
#8)CLEAR():DICTIONARY EMPTY
student={'rollno':101,'name':"dhanashree",'cname':'DS'}
student.clear()
print(student)
```

```python
# find the student with the highest marks from dictionary
students = {
    "dhanu":60,
    "shreya":70,
    "niks":65,
    "praju":80
}
highest = max(students,key=students.get)
print("Highest marks:",highest)
print("Marks",students[highest])




```

```python
#. Sort a dictionary by values using lambda.
marks={"dhanu":98,"sakshi":80,"akanksha":85,"shruti":70}
result=dict(sorted(marks.items(),key=lambda x:x[1]))
print(result)
```
